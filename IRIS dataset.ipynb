```python
import numpy as np
```

```python
data = np.genfromtxt('iris.csv', delimiter=',', skip_header=1, usecols=(0,1,2,3))
```

```python
print(f"shape:{data.shape}")
```

```python
print("mean per column:",np.mean(data,axis=0))
```

```python
print("max per column:",np.max(data,axis=0))
print("min per column:",np.min(data,axis=0))
```

```python
print("std per column:",np.std(data,axis=0))#Variance ka square root hota hai, isliye result wapas normal units mein aata hai.
#“Yeh data set stable hai ya har value random idhar udhar ja rahi hai?”
```

```python
print("var per column:",np.var(data,axis=0))#	Har value aur average ke darmiyan difference ka square karta hai
#  Variance aur Standard Deviation se pata chalta hai ke data “mean” ke around kitna idhar-udhar bhatak raha hai 
```

```python
z_score = (data - np.mean(data, axis=0)) / np.std(data, axis=0)#Sabhi features ko ek jaise scale pe lana hota hai, 
#warna kuch features jyada "weight" le lete hain
print("Z-score normalized data:\n", z_score)
```

**Standard Deviation (std):**  
Data points ka mean se kitna farq hai, yeh dikhata hai.

**Variance (var):**  
Data points ke difference ka square karke spread ko measure karta hai.

**Z-score:**  
Yeh batata hai ke data point mean se kitni doori par hai, standard deviation ke hisaab se.

```python
# Sepal Length column
sepal_length = data[:, 0]
print("Sepal Length:", sepal_length)

# First 10 flowers
first_10 = data[:10]
print("First 10 rows:\n", first_10)

# Petal Length > 1.5
petal_length_filtered = data[data[:, 2] > 1.5]
print("Filtered (Petal Length > 1.5):\n", petal_length_filtered)
```

```python
# Euclidean distance between first two rows
dist = np.linalg.norm(data[0] - data[1])
print("Euclidean distance between first two flowers:", dist)

# Count Sepal Width > mean
mean_sepal_width = np.mean(data[:, 1])
count_greater = np.sum(data[:, 1] > mean_sepal_width)
print("Count where Sepal Width > mean:", count_greater)

# Element-wise multiplication: SepalLength * PetalLength
multiplied = data[:, 0] * data[:, 2]
print("SepalLength * PetalLength:\n", multiplied)
```
